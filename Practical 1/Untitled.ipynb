```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("Employee.csv")
df.head()
```

```python
df.shape
df.columns
df.info()
df.describe()
df.isnull().sum()
```

```python
df = df.drop_duplicates()
df = df.dropna()
```

```python
df.to_csv("Employee_cleaned.csv", index=False)
```

```python
df["City"].value_counts().plot(kind="bar")
plt.title("Employees by City")
plt.xlabel("City")
plt.ylabel("Count")
plt.show()
```

```python
df["LeaveOrNot"].value_counts().plot(kind="pie", autopct="%1.1f%%")
plt.title("Leave vs Stay")
plt.ylabel("")
plt.show()
```

```python
df["Age"].plot(kind="hist", bins=10)
plt.title("Age Distribution")
plt.xlabel("Age")
plt.show()
```

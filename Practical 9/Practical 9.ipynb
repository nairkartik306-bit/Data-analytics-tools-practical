```python
# Personal information dataset
personal_data = student_data[[
    'student_id',
    'first_name',
    'last_name',
    'age',
    'email',
    'country'
]]

# Academic dataset
academic_data = student_data[[
    'student_id',
    'grade_math',
    'grade_science',
    'grade_english',
    'Final_score'
]]
```

```python
personal_data.to_csv("student_personal.csv", index=False)
academic_data.to_csv("student_academic.csv", index=False)
```

```python
# Load separate datasets
personal = pd.read_csv("student_personal.csv")
academic = pd.read_csv("student_academic.csv")
```

```python
personal.head()
```

```python
academic.head()
```

```python
personal.info()
```

```python
academic.info()
```

```python
merged_data = pd.merge(
    personal,
    academic,
    on='student_id'
)
```

```python
merged_data.head()
```

```python
merged_data.shape
```

```python
merged_data['Total_Marks'] = (
    merged_data['grade_math'] +
    merged_data['grade_science'] +
    merged_data['grade_english']
)
```

```python
merged_data['Average_Marks'] = (
    merged_data['Total_Marks'] / 3
)
```

```python
merged_data.head()
```

```python
merged_data.to_csv(
    'student_complete_data.csv',
    index=False
)
```

```python
merged_data.sort_values(
    by='Final_score',
    ascending=False
).head(10)
```

```python
print("Average Math Score:", merged_data['grade_math'].mean())
print("Average Science Score:", merged_data['grade_science'].mean())
print("Average English Score:", merged_data['grade_english'].mean())
```

```python
# Import libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Ignore warnings
import warnings
warnings.filterwarnings('ignore')
```

```python
# Load dataset
file_path = 'student.csv'

data = pd.read_csv(file_path)

# Display first 5 rows
print("First 5 Rows of Dataset")
data.head()
```

```python
# Number of rows and columns
print("Dataset Shape:")
print(data.shape)
```

```python
# Display column names
print("Column Names:")
print(data.columns)
```

```python
# Dataset information
print("Dataset Information")
data.info()
```

```python
# Statistical summary
print("Statistical Summary")
data.describe()
```

```python
# Check missing values
print("Missing Values")
print(data.isnull().sum())
```

```python
# Fill missing numerical values with mean
numeric_columns = data.select_dtypes(include=np.number).columns

for col in numeric_columns:
    data[col].fillna(data[col].mean(), inplace=True)

print("Missing Values After Cleaning")
print(data.isnull().sum())
```

```python
# Check duplicate rows
print("Duplicate Rows:")
print(data.duplicated().sum())
```

```python
# Remove duplicates
data.drop_duplicates(inplace=True)

print("Dataset Shape After Removing Duplicates:")
print(data.shape)
```

```python
# Calculate average score
data['average_score'] = (
    data['grade_math'] +
    data['grade_science'] +
    data['grade_english']
) / 3

# Display dataset
print("Dataset with Average Score")
data.head()
```

```python
# Top performing students
top_students = data.sort_values(by='average_score', ascending=False)

print("Top 10 Students")
top_students.head(10)
```

```python
# Count students by country
country_count = data['country'].value_counts()

print(country_count)
```

```python
plt.figure(figsize=(8,5))

plt.hist(data['average_score'], bins=10)

plt.title('Distribution of Average Scores')
plt.xlabel('Average Score')
plt.ylabel('Number of Students')

plt.show()
```

```python
# Calculate average marks of each subject
subjects = ['grade_math', 'grade_science', 'grade_english']
subject_means = data[subjects].mean()

# Plot
plt.figure(figsize=(8,5))

subject_means.plot(kind='bar')

plt.title('Average Marks by Subject')
plt.xlabel('Subjects')
plt.ylabel('Average Marks')

plt.xticks(rotation=0)
plt.show()
```

```python
plt.figure(figsize=(10,6))

sns.boxplot(data=data[['grade_math', 'grade_science', 'grade_english']])

plt.title('Box Plot of Subject Scores')
plt.show()
```

```python
plt.figure(figsize=(8,6))

numeric_data = data.select_dtypes(include=np.number)

sns.heatmap(numeric_data.corr(), annot=True)

plt.title('Correlation Heatmap')
plt.show()
```

```python
plt.figure(figsize=(10,6))

sns.countplot(x='country', data=data)

plt.title('Students by Country')
plt.xlabel('Country')
plt.ylabel('Count')

plt.xticks(rotation=45)
plt.show()
```

```python
print("Highest Average Score")
print(data['average_score'].max())
```

```python
print("Lowest Average Score")
print(data['average_score'].min())
```

```python
print("Overall Average Score")
print(data['average_score'].mean())
```

```python
numerical_cols = _df.select_dtypes(include=np.number).columns

print("Numerical Columns:\n")
print(numerical_cols)
```

```python
for col in numerical_cols:
    plt.figure(figsize=(6,4))
    sns.histplot(_df[col], kde=True)
    plt.title(f'Distribution of {col}')
    plt.xlabel(col)
    plt.ylabel('Frequency')
    plt.show()
```

```python
for col in numerical_cols:
    plt.figure(figsize=(6,4))
    sns.boxplot(x=_df[col])
    plt.title(f'Boxplot of {col}')
    plt.show()
```

```python
for col in categorical_cols:
    print(f"\nValue Counts for {col}:\n")
    print(_df[col].value_counts())
```

```python
categorical_cols = _df.select_dtypes(include='object').columns

print("Categorical Columns:\n")
print(categorical_cols)
```

```python
for col in categorical_cols:
    print(f"\nPercentage Distribution for {col}:\n")
    print((_df[col].value_counts(normalize=True) * 100).round(2))
```

```python
for col in categorical_cols:
    plt.figure(figsize=(8,5))
    _df[col].value_counts().plot(kind='bar')
    plt.title(f'{col} Distribution')
    plt.xlabel(col)
    plt.ylabel('Count')
    plt.xticks(rotation=45)
    plt.show()
```

```python
correlation_matrix = _df.corr(numeric_only=True)

print(correlation_matrix)
```

```python
plt.figure(figsize=(10,6))

sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap='coolwarm',
    linewidths=0.5
)

plt.title('Correlation Heatmap')
plt.show()
```
